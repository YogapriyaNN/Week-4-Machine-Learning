```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
df = pd.read_csv(r"C:\Users\ADMIN\Downloads\Student Performance Factors.csv")

print("Dataset loaded successfully!")
```

```python
print("Rows and Columns:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
print("Data Types:")
print(df.dtypes)
```

```python
categorical_missing = [
    "Teacher_Quality",
    "Parental_Education_Level",
    "Distance_from_Home"
]

for column in categorical_missing:
    df[column] = df[column].fillna(df[column].mode()[0])

print("Missing values handled successfully!")
```

```python
print("Missing Values After Cleaning:")
print(df.isnull().sum())
```

```python
df["Performance"] = np.where(
    df["Exam_Score"] >= 70,
    "High",
    "Low"
)

print("Target variable created successfully!")
```

```python
print(df["Performance"].value_counts())
```

```python
features = [
    "Hours_Studied",
    "Attendance",
    "Sleep_Hours",
    "Previous_Scores",
    "Tutoring_Sessions",
    "Physical_Activity"
]

X = df[features]
y = df["Performance"]

print("Features selected:")
print(features)

print("\nX shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed!")
```

```python
print("Scaled training data shape:", X_train_scaled.shape)
print("Scaled testing data shape:", X_test_scaled.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    random_state=42,
    max_iter=1000
)

model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully!")
```

```python
y_pred = model.predict(X_test_scaled)

print("Predictions generated successfully!")
print("First 10 predictions:")
print(y_pred[:10])
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, pos_label="High")
recall = recall_score(y_test, y_pred, pos_label="High")
f1 = f1_score(y_test, y_pred, pos_label="High")

print("Model Evaluation Metrics")
print("------------------------")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-Score : {f1:.4f}")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred, labels=["Low", "High"])

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Low", "High"]
)

disp.plot()
plt.title("Confusion Matrix - Logistic Regression")
plt.show()
```

```python
print("Confusion Matrix:")
print(cm)
```

```python
from sklearn.metrics import roc_curve, roc_auc_score

y_prob = model.predict_proba(X_test_scaled)[:, 1]

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_prob,
    pos_label="High"
)

auc_score = roc_auc_score(
    y_test,
    y_prob
)

print(f"ROC-AUC Score: {auc_score:.4f}")
```

```python
y_train_pred = model.predict(X_train_scaled)

train_accuracy = accuracy_score(y_train, y_train_pred)
test_accuracy = accuracy_score(y_test, y_pred)

print(f"Training Accuracy: {train_accuracy:.4f}")
print(f"Testing Accuracy : {test_accuracy:.4f}")
```

```python
from sklearn.metrics import classification_report

print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
print("Training Set Class Distribution:")
print(y_train.value_counts())

print("\nTesting Set Class Distribution:")
print(y_test.value_counts())
```

```python
metrics = {
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1-Score": f1
}

plt.figure(figsize=(8, 5))
plt.bar(metrics.keys(), metrics.values())

plt.ylim(0, 1)
plt.ylabel("Score")
plt.title("Logistic Regression Performance Metrics")

for i, value in enumerate(metrics.values()):
    plt.text(i, value + 0.02, f"{value:.2f}", ha="center")

plt.show()
```
